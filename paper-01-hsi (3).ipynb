```python
# =====================================================================
# R1 — LEAKAGE-AWARE SPECTRAL–SPATIAL DEEP LEARNING
#      FOR CHOLANGIOCARCINOMA DIAGNOSIS FROM MICROSCOPIC
#      HYPERSPECTRAL IMAGES
#
# EXPERIMENTS:
#   E1 = Image-wise / random split
#   E2 = Patient-wise / group-safe split
#
# DATASET:
#   Microscopic Hyperspectral Choledoch Dataset
#
# PLATFORM:
#   Kaggle GPU
#
# IMPORTANT:
#   The code streams HSI data and DOES NOT load the full ~45 GB dataset.
# =====================================================================

import os, re, gc, json, time, random, warnings
from pathlib import Path
from collections import Counter, defaultdict

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, GroupShuffleSplit
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix, classification_report,
    roc_auc_score
)

# ---------------------------------------------------------------------
# 0. REPRODUCIBILITY
# ---------------------------------------------------------------------

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

try:
    import tensorflow as tf
except Exception:
    !pip -q install tensorflow
    import tensorflow as tf

tf.random.set_seed(SEED)

print("="*80)
print("R1 — LEAKAGE-AWARE HYPERSPECTRAL CHOLANGIOCARCINOMA EXPERIMENT")
print("="*80)
print("TensorFlow:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))

# GPU memory growth
for gpu in tf.config.list_physical_devices("GPU"):
    try:
        tf.config.experimental.set_memory_growth(gpu, True)
    except:
        pass

# ---------------------------------------------------------------------
# 1. CONFIGURATION
# ---------------------------------------------------------------------

KAGGLE_INPUT = Path("/kaggle/input")
OUT = Path("/kaggle/working/R1_Leakage_Aware_HSI")
OUT.mkdir(parents=True, exist_ok=True)

# Conservative values for Kaggle
PATCH = 25

# Number of spectral bands retained after uniform spectral sampling
N_BANDS = 32

PATCHES_PER_SCENE = 40

MAX_SCENES_PER_CLASS = None       # None = use everything discovered
MAX_PATCHES_TOTAL = 12000         # controls training size

BATCH = 32
EPOCHS = 25

VAL_SIZE = 0.15
TEST_SIZE = 0.15

CLASSES = ["L", "N", "P"]
CLASS_TO_ID = {c:i for i,c in enumerate(CLASSES)}

print("\nOutput:", OUT)

# ---------------------------------------------------------------------
# 2. FIND DATASET
# ---------------------------------------------------------------------

print("\n" + "="*80)
print("STEP 1 — SEARCHING KAGGLE INPUT")
print("="*80)

roots = [p for p in KAGGLE_INPUT.iterdir() if p.is_dir()]

for r in roots:
    print("Found:", r)

# choose likely hyperspectral/choledoch dataset
keywords = ["choledoch", "hyperspectral", "cholangi", "cancer"]

DATA_ROOT = None

for r in roots:
    name = r.name.lower()
    if any(k in name for k in keywords):
        DATA_ROOT = r
        break

if DATA_ROOT is None:
    if len(roots) == 1:
        DATA_ROOT = roots[0]
    else:
        print("\nCould not uniquely identify dataset.")
        print("Available folders:")
        for r in roots:
            print(r)
        raise RuntimeError(
            "Set DATA_ROOT manually to the Microscopic Hyperspectral "
            "Choledoch dataset folder."
        )

print("\nSelected dataset:")
print(DATA_ROOT)

# ---------------------------------------------------------------------
# 3. INVENTORY FILES
# ---------------------------------------------------------------------

print("\n" + "="*80)
print("STEP 2 — DATASET INVENTORY")
print("="*80)

extensions = Counter()

all_files = []

for p in DATA_ROOT.rglob("*"):
    if p.is_file():
        extensions[p.suffix.lower()] += 1
        all_files.append(p)

print("\nTotal files:", len(all_files))

print("\nTop file extensions:")
for ext,n in extensions.most_common(25):
    print(f"{ext or '[none]':12s} {n}")

# save inventory
pd.DataFrame(
    [{"extension":k, "count":v} for k,v in extensions.items()]
).to_csv(OUT/"file_extensions.csv", index=False)

# ---------------------------------------------------------------------
# 4. HSI FILE DISCOVERY
# ---------------------------------------------------------------------

# Common hyperspectral formats
HSI_EXTS = {
    ".mat", ".npy", ".npz",
    ".hdr", ".raw", ".dat",
    ".tif", ".tiff"
}

candidate_files = [
    p for p in all_files
    if p.suffix.lower() in HSI_EXTS
]

print("\nCandidate HSI files:", len(candidate_files))

# ---------------------------------------------------------------------
# 5. LABEL DETECTION
# ---------------------------------------------------------------------

def infer_label(path):

    parts = [x.upper() for x in path.parts]

    # Strong folder-based matching first
    for part in reversed(parts):

        if part == "L":
            return "L"

        if part == "N":
            return "N"

        if part == "P":
            return "P"

    name = path.stem.upper()

    # filename prefixes such as L001, N003, P010
    m = re.match(r"^([LNP])[_\-]?\d+", name)

    if m:
        return m.group(1)

    # standalone tokens
    tokens = re.split(r"[_\-\s]+", name)

    for t in tokens:
        if t in CLASSES:
            return t

    return None


# ---------------------------------------------------------------------
# 6. PATIENT/GROUP ID DETECTION
# ---------------------------------------------------------------------

def infer_patient_id(path):

    """
    Conservative heuristic.

    Attempts to find a stable patient/sample identifier from:
      - parent folders
      - filename prefixes

    Exact split manifest is saved so it can be audited.
    """

    rel = path.relative_to(DATA_ROOT)

    parts = list(rel.parts[:-1])

    # Search parent folders for numeric IDs
    for part in reversed(parts):

        m = re.search(
            r"(?:patient|subject|case|sample|person|id)[_\-\s]*(\d+)",
            part,
            re.I
        )

        if m:
            return "PAT_" + m.group(1)

    # generic folder containing digits
    for part in reversed(parts):

        nums = re.findall(r"\d+", part)

        if nums:
            return "PAT_" + nums[0]

    # filename patterns
    stem = path.stem

    m = re.search(
        r"(?:patient|subject|case|sample|person|id)[_\-\s]*(\d+)",
        stem,
        re.I
    )

    if m:
        return "PAT_" + m.group(1)

    # first numeric sequence
    nums = re.findall(r"\d+", stem)

    if nums:
        return "PAT_" + nums[0]

    # fallback = parent directory
    return "PAT_" + str(path.parent.name)


# ---------------------------------------------------------------------
# 7. BUILD SCENE TABLE
# ---------------------------------------------------------------------

records = []

for p in candidate_files:

    label = infer_label(p)

    if label is None:
        continue

    patient = infer_patient_id(p)

    records.append({
        "path": str(p),
        "label": label,
        "patient": patient,
        "extension": p.suffix.lower(),
        "size_mb": p.stat().st_size / 1024**2
    })

scenes = pd.DataFrame(records)

print("\nLabelled candidate scenes:", len(scenes))

if len(scenes) == 0:
    raise RuntimeError(
        "No L/N/P-labelled HSI files detected. "
        "Inspect /kaggle/input directory structure and adjust infer_label()."
    )

print("\nClass distribution:")
print(scenes["label"].value_counts())

print("\nEstimated patient/groups:")
print(scenes["patient"].nunique())

print("\nExamples:")
print(scenes.head(20).to_string(index=False))

scenes.to_csv(OUT/"scene_manifest.csv", index=False)

# Optional cap per class
if MAX_SCENES_PER_CLASS is not None:

    scenes = (
        scenes.groupby("label", group_keys=False)
        .apply(
            lambda x: x.sample(
                min(len(x), MAX_SCENES_PER_CLASS),
                random_state=SEED
            )
        )
        .reset_index(drop=True)
    )

# ---------------------------------------------------------------------
# 8. LOAD DIFFERENT HSI FORMATS
# ---------------------------------------------------------------------

def orient_cube(arr):

    arr = np.asarray(arr)

    arr = np.squeeze(arr)

    if arr.ndim != 3:
        raise ValueError(f"Expected 3-D cube, received {arr.shape}")

    # spectral axis usually smallest dimension
    spectral_axis = int(np.argmin(arr.shape))

    if spectral_axis == 0:
        arr = np.transpose(arr, (1,2,0))

    elif spectral_axis == 1:
        arr = np.transpose(arr, (0,2,1))

    return arr


def load_mat(path):

    import scipy.io as sio

    try:
        data = sio.loadmat(path)

        candidates = []

        for k,v in data.items():

            if k.startswith("__"):
                continue

            if isinstance(v, np.ndarray) and v.ndim == 3:
                candidates.append(v)

        if not candidates:
            raise ValueError("No 3D array in MAT")

        # largest 3D object
        arr = max(candidates, key=lambda x:x.size)

        return orient_cube(arr)

    except NotImplementedError:

        import h5py

        with h5py.File(path, "r") as f:

            candidates = []

            def visit(name,obj):

                if hasattr(obj, "shape") and len(obj.shape)==3:
                    candidates.append(name)

            f.visititems(visit)

            if not candidates:
                raise ValueError("No 3D dataset in HDF5 MAT")

            name = max(
                candidates,
                key=lambda n: np.prod(f[n].shape)
            )

            arr = np.array(f[name])

        return orient_cube(arr)


def load_envi(hdr_path):

    try:
        from spectral import open_image
    except:
        !pip -q install spectral
        from spectral import open_image

    img = open_image(str(hdr_path))

    return np.asarray(img.open_memmap())


def find_hdr_for_binary(path):

    candidates = [
        path.with_suffix(".hdr"),
        Path(str(path) + ".hdr"),
        path.parent / (path.stem + ".hdr")
    ]

    for x in candidates:
        if x.exists():
            return x

    return None


def load_cube(path):

    path = Path(path)

    ext = path.suffix.lower()

    if ext == ".npy":
        return orient_cube(np.load(path, mmap_mode="r"))

    if ext == ".npz":

        z = np.load(path)

        keys = list(z.keys())

        arrays = [
            z[k] for k in keys
            if np.asarray(z[k]).ndim == 3
        ]

        if not arrays:
            raise ValueError("No 3D NPZ array")

        return orient_cube(max(arrays,key=lambda x:x.size))

    if ext == ".mat":
        return load_mat(path)

    if ext == ".hdr":
        return load_envi(path)

    if ext in [".raw",".dat"]:

        hdr = find_hdr_for_binary(path)

        if hdr:
            return load_envi(hdr)

        raise ValueError("RAW/DAT has no matching HDR")

    if ext in [".tif",".tiff"]:

        import tifffile

        return orient_cube(tifffile.imread(path))

    raise ValueError("Unsupported format")


# ---------------------------------------------------------------------
# 9. TEST READ SAMPLE FILES
# ---------------------------------------------------------------------

print("\n" + "="*80)
print("STEP 3 — TESTING HSI READERS")
print("="*80)

valid_records = []

seen_hdr_targets = set()

for _,row in scenes.iterrows():

    p = Path(row.path)

    # Avoid loading binary file separately if corresponding HDR exists
    if p.suffix.lower() in [".raw",".dat"]:

        hdr = find_hdr_for_binary(p)

        if hdr and str(hdr) in seen_hdr_targets:
            continue

    try:

        cube = load_cube(p)

        h,w,b = cube.shape

        if h < PATCH or w < PATCH or b < 3:
            continue

        valid_records.append({
            **row.to_dict(),
            "height":h,
            "width":w,
            "bands":b
        })

        if p.suffix.lower()==".hdr":
            seen_hdr_targets.add(str(p))

        del cube

    except Exception as e:

        pass

    if len(valid_records) % 50 == 0 and len(valid_records)>0:
        print("Validated:", len(valid_records))

scenes = pd.DataFrame(valid_records)

if len(scenes)==0:
    raise RuntimeError(
        "Files were detected but no HSI cubes could be read. "
        "Review file_extensions.csv and scene_manifest.csv."
    )

scenes.to_csv(OUT/"validated_scenes.csv",index=False)

print("\nReadable HSI scenes:",len(scenes))

print("\nBands:")
print(scenes["bands"].describe())

print("\nClass distribution:")
print(scenes.label.value_counts())

print("\nPatients/groups:", scenes.patient.nunique())

print("\nExample dimensions:")
print(
    scenes[
        ["label","patient","height","width","bands","path"]
    ].head(15).to_string(index=False)
)

# ---------------------------------------------------------------------
# 10. SPECTRAL BAND REDUCTION
# ---------------------------------------------------------------------

def spectral_reduce(cube, target=N_BANDS):

    b = cube.shape[-1]

    if b <= target:
        return np.asarray(cube,dtype=np.float32)

    idx = np.linspace(
        0,
        b-1,
        target
    ).astype(int)

    return np.asarray(
        cube[:,:,idx],
        dtype=np.float32
    )


# ---------------------------------------------------------------------
# 11. ROBUST NORMALIZATION
# ---------------------------------------------------------------------

def normalize_cube(cube):

    cube = np.nan_to_num(
        cube.astype(np.float32),
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    # Per-band robust normalization
    flat = cube.reshape(-1,cube.shape[-1])

    # sample pixels to reduce computation
    if len(flat)>10000:

        idx=np.random.choice(
            len(flat),
            10000,
            replace=False
        )

        sample=flat[idx]

    else:
        sample=flat

    lo=np.percentile(sample,1,axis=0)
    hi=np.percentile(sample,99,axis=0)

    scale=np.maximum(hi-lo,1e-6)

    cube=(cube-lo)/scale

    return np.clip(cube,0,1)


# ---------------------------------------------------------------------
# 12. PATCH EXTRACTION
# ---------------------------------------------------------------------

def extract_patches(cube,n_patches):

    h,w,b=cube.shape

    half=PATCH//2

    patches=[]

    attempts=0

    while len(patches)<n_patches and attempts<n_patches*10:

        attempts+=1

        y=np.random.randint(half,h-half)
        x=np.random.randint(half,w-half)

        patch=cube[
            y-half:y+half+1,
            x-half:x+half+1,
            :
        ]

        if patch.shape!=(PATCH,PATCH,b):
            continue

        # reject near-empty patches
        if np.std(patch)<1e-4:
            continue

        patches.append(patch.astype(np.float16))

    return patches


# ---------------------------------------------------------------------
# 13. CREATE PATCH CACHE
# ---------------------------------------------------------------------

print("\n" + "="*80)
print("STEP 4 — EXTRACTING SPECTRAL–SPATIAL PATCHES")
print("="*80)

X=[]
Y=[]
GROUPS=[]
SCENE_IDS=[]

order=scenes.sample(frac=1,random_state=SEED)

for count,(_,row) in enumerate(order.iterrows(),1):

    if len(X)>=MAX_PATCHES_TOTAL:
        break

    try:

        cube=load_cube(row.path)

        cube=spectral_reduce(cube,N_BANDS)

        cube=normalize_cube(cube)

        patches=extract_patches(
            cube,
            PATCHES_PER_SCENE
        )

        remaining=MAX_PATCHES_TOTAL-len(X)

        patches=patches[:remaining]

        X.extend(patches)

        Y.extend(
            [CLASS_TO_ID[row.label]]*len(patches)
        )

        GROUPS.extend(
            [row.patient]*len(patches)
        )

        SCENE_IDS.extend(
            [row.path]*len(patches)
        )

        del cube,patches

        gc.collect()

        if count%10==0:

            print(
                f"Scenes processed: {count}/{len(order)} | "
                f"Patches: {len(X)}"
            )

    except Exception as e:

        print(
            "Skipped:",
            Path(row.path).name,
            "|",
            str(e)[:100]
        )


X=np.asarray(X,dtype=np.float16)
Y=np.asarray(Y,dtype=np.int32)
GROUPS=np.asarray(GROUPS)
SCENE_IDS=np.asarray(SCENE_IDS)

print("\nPatch dataset:",X.shape)
print("Labels:",Counter(Y))
print("Groups:",len(np.unique(GROUPS)))

np.save(OUT/"patch_labels.npy",Y)
np.save(OUT/"patch_groups.npy",GROUPS)

pd.DataFrame({
    "scene":SCENE_IDS,
    "patient":GROUPS,
    "label":[CLASSES[x] for x in Y]
}).to_csv(
    OUT/"patch_manifest.csv",
    index=False
)

# ---------------------------------------------------------------------
# 14. SPLIT A — IMAGE/PATCH-WISE RANDOM SPLIT
# ---------------------------------------------------------------------

print("\n" + "="*80)
print("STEP 5 — IMAGE-WISE RANDOM SPLIT")
print("="*80)

idx=np.arange(len(Y))

train_idx,temp_idx=train_test_split(
    idx,
    test_size=VAL_SIZE+TEST_SIZE,
    stratify=Y,
    random_state=SEED
)

relative_test=TEST_SIZE/(VAL_SIZE+TEST_SIZE)

val_idx,test_idx=train_test_split(
    temp_idx,
    test_size=relative_test,
    stratify=Y[temp_idx],
    random_state=SEED
)

random_split={
    "train":train_idx,
    "val":val_idx,
    "test":test_idx
}

# ---------------------------------------------------------------------
# 15. SPLIT B — PATIENT-WISE SPLIT
# ---------------------------------------------------------------------

print("\n" + "="*80)
print("STEP 6 — PATIENT-WISE LEAKAGE-SAFE SPLIT")
print("="*80)

gss1=GroupShuffleSplit(
    n_splits=1,
    test_size=VAL_SIZE+TEST_SIZE,
    random_state=SEED
)

train_g,temp_g=next(
    gss1.split(
        idx,
        Y,
        groups=GROUPS
    )
)

gss2=GroupShuffleSplit(
    n_splits=1,
    test_size=0.5,
    random_state=SEED+1
)

val_rel,test_rel=next(
    gss2.split(
        temp_g,
        Y[temp_g],
        groups=GROUPS[temp_g]
    )
)

val_g=temp_g[val_rel]
test_g=temp_g[test_rel]

patient_split={
    "train":train_g,
    "val":val_g,
    "test":test_g
}

# ---------------------------------------------------------------------
# 16. LEAKAGE AUDIT
# ---------------------------------------------------------------------

def audit_split(split,name):

    tr=set(GROUPS[split["train"]])
    va=set(GROUPS[split["val"]])
    te=set(GROUPS[split["test"]])

    audit={
        "experiment":name,

        "train_samples":
            len(split["train"]),

        "val_samples":
            len(split["val"]),

        "test_samples":
            len(split["test"]),

        "train_patients":
            len(tr),

        "val_patients":
            len(va),

        "test_patients":
            len(te),

        "train_val_patient_overlap":
            len(tr & va),

        "train_test_patient_overlap":
            len(tr & te),

        "val_test_patient_overlap":
            len(va & te)
    }

    return audit


audit_random=audit_split(
    random_split,
    "Image-wise"
)

audit_patient=audit_split(
    patient_split,
    "Patient-wise"
)

audit_df=pd.DataFrame([
    audit_random,
    audit_patient
])

print("\nLEAKAGE AUDIT")
print(audit_df.to_string(index=False))

audit_df.to_csv(
    OUT/"leakage_audit.csv",
    index=False
)

# ---------------------------------------------------------------------
# 17. SAVE EXACT SPLITS
# ---------------------------------------------------------------------

def save_split(split,name):

    rows=[]

    for subset,indices in split.items():

        for i in indices:

            rows.append({
                "index":int(i),
                "subset":subset,
                "label":CLASSES[Y[i]],
                "patient":GROUPS[i],
                "scene":SCENE_IDS[i]
            })

    pd.DataFrame(rows).to_csv(
        OUT/f"{name}_split_manifest.csv",
        index=False
    )


save_split(
    random_split,
    "imagewise"
)

save_split(
    patient_split,
    "patientwise"
)

# ---------------------------------------------------------------------
# 18. 3D SPECTRAL–SPATIAL CNN
# ---------------------------------------------------------------------

from tensorflow.keras import layers,models

def build_model(input_shape,n_classes):

    inp=layers.Input(shape=input_shape)

    # Add channel dimension:
    # H x W x Bands x 1
    x=layers.Reshape(
        (
            input_shape[0],
            input_shape[1],
            input_shape[2],
            1
        )
    )(inp)

    x=layers.Conv3D(
        16,
        (3,3,7),
        padding="same",
        activation="relu"
    )(x)

    x=layers.BatchNormalization()(x)

    x=layers.MaxPool3D(
        (2,2,2)
    )(x)

    x=layers.Conv3D(
        32,
        (3,3,5),
        padding="same",
        activation="relu"
    )(x)

    x=layers.BatchNormalization()(x)

    x=layers.MaxPool3D(
        (2,2,2)
    )(x)

    x=layers.Conv3D(
        64,
        (3,3,3),
        padding="same",
        activation="relu"
    )(x)

    x=layers.BatchNormalization()(x)

    x=layers.GlobalAveragePooling3D()(x)

    x=layers.Dense(
        128,
        activation="relu"
    )(x)

    x=layers.Dropout(0.4)(x)

    out=layers.Dense(
        n_classes,
        activation="softmax"
    )(x)

    model=models.Model(inp,out)

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            1e-3
        ),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ---------------------------------------------------------------------
# 19. TRAIN/EVALUATE FUNCTION
# ---------------------------------------------------------------------

def run_experiment(split,name):

    print("\n")
    print("="*80)
    print("TRAINING:",name)
    print("="*80)

    tf.keras.backend.clear_session()

    model=build_model(
        X.shape[1:],
        len(CLASSES)
    )

    model.summary()

    callbacks=[
        tf.keras.callbacks.EarlyStopping(
            monitor="val_loss",
            patience=5,
            restore_best_weights=True
        ),

        tf.keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss",
            patience=2,
            factor=0.5,
            min_lr=1e-6
        ),

        tf.keras.callbacks.ModelCheckpoint(
            OUT/f"{name}_best.keras",
            monitor="val_loss",
            save_best_only=True
        )
    ]

    history=model.fit(
        X[split["train"]].astype(np.float32),
        Y[split["train"]],

        validation_data=(
            X[split["val"]].astype(np.float32),
            Y[split["val"]]
        ),

        epochs=EPOCHS,
        batch_size=BATCH,
        callbacks=callbacks,
        verbose=1
    )

    test_indices=split["test"]

    probs=model.predict(
        X[test_indices].astype(np.float32),
        batch_size=BATCH,
        verbose=0
    )

    pred=np.argmax(probs,axis=1)

    true=Y[test_indices]

    acc=accuracy_score(
        true,
        pred
    )

    bacc=balanced_accuracy_score(
        true,
        pred
    )

    precision,recall,f1,_=precision_recall_fscore_support(
        true,
        pred,
        average="macro",
        zero_division=0
    )

    cm=confusion_matrix(
        true,
        pred,
        labels=range(len(CLASSES))
    )

    results={
        "experiment":name,
        "accuracy":acc,
        "balanced_accuracy":bacc,
        "macro_precision":precision,
        "macro_recall":recall,
        "macro_f1":f1,
        "test_samples":len(true)
    }

    print("\nRESULT")
    print(results)

    print("\nClassification report:")
    print(
        classification_report(
            true,
            pred,
            target_names=CLASSES,
            zero_division=0
        )
    )

    # Predictions
    pred_df=pd.DataFrame({
        "true_id":true,
        "pred_id":pred,
        "true_label":[CLASSES[i] for i in true],
        "pred_label":[CLASSES[i] for i in pred],
        "patient":GROUPS[test_indices],
        "scene":SCENE_IDS[test_indices],
        "confidence":np.max(probs,axis=1)
    })

    for i,c in enumerate(CLASSES):
        pred_df[f"prob_{c}"]=probs[:,i]

    pred_df.to_csv(
        OUT/f"{name}_predictions.csv",
        index=False
    )

    # Confusion matrix
    plt.figure(figsize=(6,5))

    plt.imshow(cm)

    plt.xticks(
        range(len(CLASSES)),
        CLASSES
    )

    plt.yticks(
        range(len(CLASSES)),
        CLASSES
    )

    plt.xlabel("Predicted")
    plt.ylabel("True")

    plt.title(
        f"{name}\nConfusion Matrix"
    )

    for i in range(len(CLASSES)):
        for j in range(len(CLASSES)):
            plt.text(
                j,i,
                cm[i,j],
                ha="center",
                va="center"
            )

    plt.tight_layout()

    plt.savefig(
        OUT/f"{name}_confusion_matrix.png",
        dpi=300
    )

    plt.show()

    # Learning curves
    hist=pd.DataFrame(history.history)

    hist.to_csv(
        OUT/f"{name}_training_history.csv",
        index=False
    )

    plt.figure(figsize=(7,5))

    plt.plot(
        hist["accuracy"],
        label="Train"
    )

    plt.plot(
        hist["val_accuracy"],
        label="Validation"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.title(name)
    plt.legend()
    plt.tight_layout()

    plt.savefig(
        OUT/f"{name}_learning_curve.png",
        dpi=300
    )

    plt.show()

    del model
    gc.collect()

    return results


# ---------------------------------------------------------------------
# 20. RUN BOTH EXPERIMENTS
# ---------------------------------------------------------------------

result_imagewise=run_experiment(
    random_split,
    "ImageWise_3DCNN"
)

result_patientwise=run_experiment(
    patient_split,
    "PatientWise_3DCNN"
)

# ---------------------------------------------------------------------
# 21. FINAL COMPARISON
# ---------------------------------------------------------------------

results=pd.DataFrame([
    result_imagewise,
    result_patientwise
])

results["accuracy_drop_vs_imagewise"]=(
    results.iloc[0]["accuracy"]
    -
    results["accuracy"]
)

results["f1_drop_vs_imagewise"]=(
    results.iloc[0]["macro_f1"]
    -
    results["macro_f1"]
)

results.to_csv(
    OUT/"R1_FINAL_RESULTS.csv",
    index=False
)

print("\n")
print("="*80)
print("R1 FINAL COMPARISON")
print("="*80)

print(
    results.round(4).to_string(index=False)
)

# ---------------------------------------------------------------------
# 22. LEAKAGE EFFECT
# ---------------------------------------------------------------------

image_acc=result_imagewise["accuracy"]
patient_acc=result_patientwise["accuracy"]

image_f1=result_imagewise["macro_f1"]
patient_f1=result_patientwise["macro_f1"]

summary={
    "Image-wise Accuracy":float(image_acc),
    "Patient-wise Accuracy":float(patient_acc),

    "Accuracy Difference":
        float(image_acc-patient_acc),

    "Image-wise Macro F1":
        float(image_f1),

    "Patient-wise Macro F1":
        float(patient_f1),

    "Macro F1 Difference":
        float(image_f1-patient_f1)
}

with open(
    OUT/"leakage_effect_summary.json",
    "w"
) as f:

    json.dump(
        summary,
        f,
        indent=4
    )

print("\nLeakage effect:")

for k,v in summary.items():
    print(f"{k:30s}: {v:.4f}")

# ---------------------------------------------------------------------
# 23. COMPARISON FIGURE
# ---------------------------------------------------------------------

plot_df=results.set_index("experiment")[
    [
        "accuracy",
        "balanced_accuracy",
        "macro_f1"
    ]
]

plot_df.plot(
    kind="bar",
    figsize=(8,5)
)

plt.ylim(0,1)
plt.ylabel("Score")
plt.title(
    "Image-wise vs Patient-wise Evaluation"
)
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig(
    OUT/"Leakage_Comparison.png",
    dpi=300
)

plt.show()

# ---------------------------------------------------------------------
# 24. CREATE ZIP
# ---------------------------------------------------------------------

import shutil

zip_path="/kaggle/working/R1_Leakage_Aware_HSI_Results"

shutil.make_archive(
    zip_path,
    "zip",
    OUT
)

print("\n")
print("="*80)
print("R1 COMPLETE")
print("="*80)

print("Results folder:")
print(OUT)

print("\nDownload:")
print(
    "/kaggle/working/"
    "R1_Leakage_Aware_HSI_Results.zip"
)

print("\nMain files:")
print("  R1_FINAL_RESULTS.csv")
print("  leakage_audit.csv")
print("  leakage_effect_summary.json")
print("  imagewise_split_manifest.csv")
print("  patientwise_split_manifest.csv")
print("  ImageWise_3DCNN_predictions.csv")
print("  PatientWise_3DCNN_predictions.csv")
print("  ImageWise_3DCNN_confusion_matrix.png")
print("  PatientWise_3DCNN_confusion_matrix.png")
print("  Leakage_Comparison.png")

print("\n*** IMPORTANT FOR THE PAPER ***")
print(
    "Report PATIENT-WISE performance as the primary result. "
    "Image-wise performance is included to quantify possible "
    "data-leakage inflation."
)

print("="*80)
```

```python
# =============================================================================
# R2 — HYBRID QUANTUM–CLASSICAL SPECTRAL–SPATIAL LEARNING
#      FOR CHOLANGIOCARCINOMA DIAGNOSIS FROM HYPERSPECTRAL PATHOLOGY
#
# Architecture:
# HSI Cube
#   -> spectral reduction
#   -> spectral-spatial patches
#   -> 3D CNN feature extractor
#   -> PCA
#   -> quantum angle embedding
#   -> Variational Quantum Circuit (VQC)
#   -> L / N / P classification
#
# Comparisons:
#   B1: CNN + MLP
#   B2: CNN + RBF-SVM
#   Q1: CNN + PCA + VQC
#
# Evaluation:
#   PATIENT-WISE / GROUP-SAFE SPLIT ONLY
#
# Designed for:
#   Kaggle GPU + Microscopic Hyperspectral Choledoch Dataset
#
# NOTE:
#   Quantum simulation runs primarily on CPU.
#   The GPU accelerates CNN training.
# =============================================================================

import os, re, gc, json, time, random, warnings, shutil, subprocess, sys
from pathlib import Path
from collections import Counter

warnings.filterwarnings("ignore")

# =============================================================================
# 0. INSTALL / IMPORT
# =============================================================================

try:
    import pennylane as qml
except:
    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "-q", "pennylane"]
    )
    import pennylane as qml

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
    classification_report
)

import tensorflow as tf
from tensorflow.keras import layers, models

# =============================================================================
# 1. REPRODUCIBILITY
# =============================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("=" * 90)
print("R2 — HYBRID QUANTUM–CLASSICAL HYPERSPECTRAL CHOLANGIOCARCINOMA")
print("=" * 90)

print("TensorFlow :", tf.__version__)
print("PennyLane  :", qml.__version__)
print("GPU        :", tf.config.list_physical_devices("GPU"))

for gpu in tf.config.list_physical_devices("GPU"):
    try:
        tf.config.experimental.set_memory_growth(gpu, True)
    except:
        pass

# =============================================================================
# 2. CONFIGURATION
# =============================================================================

INPUT_ROOT = Path("/kaggle/input")

OUT = Path("/kaggle/working/R2_Hybrid_Quantum_HSI")
OUT.mkdir(parents=True, exist_ok=True)

# HSI settings
PATCH = 25
N_BANDS = 32

PATCHES_PER_SCENE = 30

# Keep manageable for Kaggle.
MAX_PATCHES_TOTAL = 9000

# CNN
CNN_EPOCHS = 20
BATCH = 32

# Quantum
N_QUBITS = 6

# 6 PCA features -> 6 qubits
PCA_COMPONENTS = N_QUBITS

VQC_LAYERS = 3

# Quantum simulation is expensive.
# We deliberately use a balanced feature subset.
MAX_Q_TRAIN = 1200
MAX_Q_VAL   = 300
MAX_Q_TEST  = 600

VQC_EPOCHS = 30
VQC_BATCH = 24

LR_Q = 0.02

CLASSES = ["L", "N", "P"]

CLASS_TO_ID = {
    "L": 0,
    "N": 1,
    "P": 2
}

print("\nOutput:")
print(OUT)

# =============================================================================
# 3. FIND DATASET
# =============================================================================

print("\n" + "=" * 90)
print("STEP 1 — FINDING DATASET")
print("=" * 90)

roots = [
    p for p in INPUT_ROOT.iterdir()
    if p.is_dir()
]

for r in roots:
    print(r)

keywords = [
    "choledoch",
    "hyperspectral",
    "cholangi",
    "cancer"
]

DATA_ROOT = None

for r in roots:

    if any(
        x in r.name.lower()
        for x in keywords
    ):
        DATA_ROOT = r
        break

if DATA_ROOT is None:

    if len(roots) == 1:
        DATA_ROOT = roots[0]

    else:
        raise RuntimeError(
            "Could not automatically identify dataset folder. "
            "Set DATA_ROOT manually."
        )

print("\nSelected:")
print(DATA_ROOT)

# =============================================================================
# 4. DATASET INVENTORY
# =============================================================================

print("\n" + "=" * 90)
print("STEP 2 — INVENTORY")
print("=" * 90)

all_files = [
    p for p in DATA_ROOT.rglob("*")
    if p.is_file()
]

print("Total files:", len(all_files))

ext_count = Counter(
    p.suffix.lower()
    for p in all_files
)

print("\nExtensions:")

for ext, n in ext_count.most_common(30):
    print(
        f"{ext or '[none]':15s}",
        n
    )

pd.DataFrame(
    [
        {
            "extension": k,
            "count": v
        }
        for k, v in ext_count.items()
    ]
).to_csv(
    OUT / "file_inventory.csv",
    index=False
)

# =============================================================================
# 5. LABEL AND PATIENT DETECTION
# =============================================================================

HSI_EXTS = {
    ".mat",
    ".npy",
    ".npz",
    ".hdr",
    ".raw",
    ".dat",
    ".tif",
    ".tiff"
}


def infer_label(path):

    # folders
    for part in reversed(path.parts):

        p = part.upper().strip()

        if p in CLASSES:
            return p

    # filename
    stem = path.stem.upper()

    m = re.match(
        r"^([LNP])[_\-]?\d+",
        stem
    )

    if m:
        return m.group(1)

    tokens = re.split(
        r"[_\-\s]+",
        stem
    )

    for token in tokens:

        if token in CLASSES:
            return token

    return None


def infer_patient(path):

    rel = path.relative_to(DATA_ROOT)

    # inspect folders first
    for part in reversed(rel.parts[:-1]):

        m = re.search(
            r"(?:patient|subject|case|person|sample|id)"
            r"[_\-\s]*(\d+)",
            part,
            re.I
        )

        if m:
            return "PAT_" + m.group(1)

    # numeric folder
    for part in reversed(rel.parts[:-1]):

        nums = re.findall(
            r"\d+",
            part
        )

        if nums:
            return "PAT_" + nums[0]

    # filename
    stem = path.stem

    m = re.search(
        r"(?:patient|subject|case|person|sample|id)"
        r"[_\-\s]*(\d+)",
        stem,
        re.I
    )

    if m:
        return "PAT_" + m.group(1)

    nums = re.findall(
        r"\d+",
        stem
    )

    if nums:
        return "PAT_" + nums[0]

    return "PAT_" + path.parent.name


# =============================================================================
# 6. BUILD SCENE TABLE
# =============================================================================

records = []

for p in all_files:

    if p.suffix.lower() not in HSI_EXTS:
        continue

    label = infer_label(p)

    if label is None:
        continue

    records.append(
        {
            "path": str(p),
            "label": label,
            "patient": infer_patient(p),
            "extension": p.suffix.lower(),
            "size_mb":
                p.stat().st_size / 1024**2
        }
    )

scenes = pd.DataFrame(records)

if len(scenes) == 0:

    raise RuntimeError(
        "No L/N/P-labelled HSI files detected."
    )

print("\nCandidate scenes:", len(scenes))

print("\nClasses:")
print(
    scenes.label.value_counts()
)

print(
    "\nEstimated patients:",
    scenes.patient.nunique()
)

scenes.to_csv(
    OUT / "candidate_scenes.csv",
    index=False
)

# =============================================================================
# 7. HSI LOADERS
# =============================================================================

def orient_cube(arr):

    arr = np.asarray(arr)
    arr = np.squeeze(arr)

    if arr.ndim != 3:
        raise ValueError(
            f"Not a 3-D HSI cube: {arr.shape}"
        )

    # Spectral dimension is generally the smallest.
    axis = int(
        np.argmin(arr.shape)
    )

    if axis == 0:

        arr = np.transpose(
            arr,
            (1, 2, 0)
        )

    elif axis == 1:

        arr = np.transpose(
            arr,
            (0, 2, 1)
        )

    return arr


def load_mat(path):

    import scipy.io as sio

    try:

        data = sio.loadmat(path)

        arrays = [
            v for k, v in data.items()
            if not k.startswith("__")
            and isinstance(v, np.ndarray)
            and v.ndim == 3
        ]

        if not arrays:
            raise ValueError(
                "No 3D array"
            )

        return orient_cube(
            max(
                arrays,
                key=lambda x: x.size
            )
        )

    except NotImplementedError:

        import h5py

        with h5py.File(
            path,
            "r"
        ) as f:

            names = []

            def visit(name, obj):

                if (
                    hasattr(obj, "shape")
                    and len(obj.shape) == 3
                ):
                    names.append(name)

            f.visititems(visit)

            if not names:
                raise ValueError(
                    "No HDF5 3D array"
                )

            best = max(
                names,
                key=lambda n:
                    np.prod(f[n].shape)
            )

            arr = np.array(
                f[best]
            )

        return orient_cube(arr)


def load_envi(hdr):

    try:
        from spectral import open_image

    except:
        subprocess.check_call(
            [
                sys.executable,
                "-m",
                "pip",
                "install",
                "-q",
                "spectral"
            ]
        )

        from spectral import open_image

    img = open_image(
        str(hdr)
    )

    return np.asarray(
        img.open_memmap()
    )


def find_hdr(path):

    possibilities = [
        path.with_suffix(".hdr"),
        Path(str(path) + ".hdr"),
        path.parent /
        (path.stem + ".hdr")
    ]

    for x in possibilities:

        if x.exists():
            return x

    return None


def load_cube(path):

    path = Path(path)

    ext = path.suffix.lower()

    if ext == ".mat":

        return load_mat(path)

    elif ext == ".npy":

        return orient_cube(
            np.load(
                path,
                mmap_mode="r"
            )
        )

    elif ext == ".npz":

        z = np.load(path)

        arrays = [
            z[k]
            for k in z.keys()
            if z[k].ndim == 3
        ]

        if not arrays:
            raise ValueError(
                "No 3D NPZ array"
            )

        return orient_cube(
            max(
                arrays,
                key=lambda x: x.size
            )
        )

    elif ext == ".hdr":

        return load_envi(path)

    elif ext in [
        ".raw",
        ".dat"
    ]:

        hdr = find_hdr(path)

        if hdr is None:
            raise ValueError(
                "No HDR file"
            )

        return load_envi(hdr)

    elif ext in [
        ".tif",
        ".tiff"
    ]:

        import tifffile

        return orient_cube(
            tifffile.imread(path)
        )

    raise ValueError(
        "Unsupported format"
    )


# =============================================================================
# 8. VALIDATE HSI FILES
# =============================================================================

print("\n" + "=" * 90)
print("STEP 3 — VALIDATING HSI CUBES")
print("=" * 90)

valid = []

processed_hdr = set()

for i, row in scenes.iterrows():

    p = Path(row.path)

    # Prevent RAW/HDR duplication
    if p.suffix.lower() in [
        ".raw",
        ".dat"
    ]:

        hdr = find_hdr(p)

        if hdr and str(hdr) in processed_hdr:
            continue

    try:

        cube = load_cube(p)

        h, w, b = cube.shape

        if (
            h >= PATCH
            and w >= PATCH
            and b >= 4
        ):

            valid.append(
                {
                    **row.to_dict(),
                    "height": h,
                    "width": w,
                    "bands": b
                }
            )

        if p.suffix.lower() == ".hdr":
            processed_hdr.add(
                str(p)
            )

        del cube

    except:
        pass

    if (
        len(valid) > 0
        and len(valid) % 50 == 0
    ):

        print(
            "Validated:",
            len(valid)
        )

scenes = pd.DataFrame(valid)

if len(scenes) == 0:

    raise RuntimeError(
        "HSI files found but none could be read."
    )

print(
    "\nReadable HSI scenes:",
    len(scenes)
)

print("\nClass distribution:")

print(
    scenes.label.value_counts()
)

print(
    "\nPatients:",
    scenes.patient.nunique()
)

print("\nBand statistics:")

print(
    scenes.bands.describe()
)

scenes.to_csv(
    OUT / "validated_scenes.csv",
    index=False
)

# =============================================================================
# 9. SPECTRAL REDUCTION
# =============================================================================

def spectral_reduce(cube):

    bands = cube.shape[-1]

    if bands <= N_BANDS:

        return np.asarray(
            cube,
            dtype=np.float32
        )

    indices = np.linspace(
        0,
        bands - 1,
        N_BANDS
    ).astype(int)

    return np.asarray(
        cube[:, :, indices],
        dtype=np.float32
    )


# =============================================================================
# 10. NORMALIZATION
# =============================================================================

def normalize_cube(cube):

    cube = np.nan_to_num(
        cube.astype(np.float32),
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    flat = cube.reshape(
        -1,
        cube.shape[-1]
    )

    if len(flat) > 10000:

        ids = np.random.choice(
            len(flat),
            10000,
            replace=False
        )

        sample = flat[ids]

    else:

        sample = flat

    low = np.percentile(
        sample,
        1,
        axis=0
    )

    high = np.percentile(
        sample,
        99,
        axis=0
    )

    scale = np.maximum(
        high - low,
        1e-6
    )

    cube = (
        cube - low
    ) / scale

    return np.clip(
        cube,
        0,
        1
    )


# =============================================================================
# 11. PATCH EXTRACTION
# =============================================================================

def extract_patches(
    cube,
    n
):

    h, w, b = cube.shape

    half = PATCH // 2

    result = []

    attempts = 0

    while (
        len(result) < n
        and attempts < n * 12
    ):

        attempts += 1

        y = np.random.randint(
            half,
            h - half
        )

        x = np.random.randint(
            half,
            w - half
        )

        patch = cube[
            y-half:y+half+1,
            x-half:x+half+1,
            :
        ]

        if patch.shape != (
            PATCH,
            PATCH,
            b
        ):
            continue

        if np.std(patch) < 1e-4:
            continue

        result.append(
            patch.astype(
                np.float16
            )
        )

    return result


# =============================================================================
# 12. BUILD PATCH DATASET
# =============================================================================

print("\n" + "=" * 90)
print("STEP 4 — BUILDING HSI PATCH DATASET")
print("=" * 90)

X = []
Y = []
GROUP = []
SCENE = []

ordered = scenes.sample(
    frac=1,
    random_state=SEED
)

for n, (_, row) in enumerate(
    ordered.iterrows(),
    1
):

    if len(X) >= MAX_PATCHES_TOTAL:
        break

    try:

        cube = load_cube(
            row.path
        )

        cube = spectral_reduce(
            cube
        )

        cube = normalize_cube(
            cube
        )

        patches = extract_patches(
            cube,
            PATCHES_PER_SCENE
        )

        remaining = (
            MAX_PATCHES_TOTAL
            - len(X)
        )

        patches = patches[
            :remaining
        ]

        X.extend(
            patches
        )

        Y.extend(
            [
                CLASS_TO_ID[
                    row.label
                ]
            ] * len(patches)
        )

        GROUP.extend(
            [
                row.patient
            ] * len(patches)
        )

        SCENE.extend(
            [
                row.path
            ] * len(patches)
        )

        del cube, patches

        gc.collect()

        if n % 10 == 0:

            print(
                f"Scenes {n}/{len(ordered)} | "
                f"Patches {len(X)}"
            )

    except Exception as e:

        print(
            "Skipped:",
            Path(row.path).name,
            str(e)[:80]
        )


X = np.asarray(
    X,
    dtype=np.float16
)

Y = np.asarray(
    Y,
    dtype=np.int32
)

GROUP = np.asarray(
    GROUP
)

SCENE = np.asarray(
    SCENE
)

print(
    "\nPatch tensor:",
    X.shape
)

print(
    "Class distribution:",
    Counter(Y)
)

print(
    "Patients:",
    len(np.unique(GROUP))
)

# =============================================================================
# 13. PATIENT-WISE SPLIT
# =============================================================================

print("\n" + "=" * 90)
print("STEP 5 — PATIENT-WISE SPLIT")
print("=" * 90)

indices = np.arange(
    len(Y)
)

split1 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.30,
    random_state=SEED
)

train_idx, temp_idx = next(
    split1.split(
        indices,
        Y,
        groups=GROUP
    )
)

split2 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=SEED + 1
)

val_rel, test_rel = next(
    split2.split(
        temp_idx,
        Y[temp_idx],
        groups=GROUP[temp_idx]
    )
)

val_idx = temp_idx[
    val_rel
]

test_idx = temp_idx[
    test_rel
]

# ---------------------------------------------------------------------
# Leakage audit
# ---------------------------------------------------------------------

train_patients = set(
    GROUP[train_idx]
)

val_patients = set(
    GROUP[val_idx]
)

test_patients = set(
    GROUP[test_idx]
)

print(
    "Train patches:",
    len(train_idx)
)

print(
    "Validation patches:",
    len(val_idx)
)

print(
    "Test patches:",
    len(test_idx)
)

print()

print(
    "Train patients:",
    len(train_patients)
)

print(
    "Validation patients:",
    len(val_patients)
)

print(
    "Test patients:",
    len(test_patients)
)

print()

print(
    "Train/Val overlap:",
    len(
        train_patients
        & val_patients
    )
)

print(
    "Train/Test overlap:",
    len(
        train_patients
        & test_patients
    )
)

print(
    "Val/Test overlap:",
    len(
        val_patients
        & test_patients
    )
)

assert len(
    train_patients
    & val_patients
) == 0

assert len(
    train_patients
    & test_patients
) == 0

assert len(
    val_patients
    & test_patients
) == 0

print(
    "\n✓ PATIENT-LEVEL LEAKAGE = ZERO"
)

# Save manifest
manifest = []

for subset, ids in [
    ("train", train_idx),
    ("validation", val_idx),
    ("test", test_idx)
]:

    for i in ids:

        manifest.append(
            {
                "index": int(i),
                "subset": subset,
                "class":
                    CLASSES[Y[i]],
                "patient":
                    GROUP[i],
                "scene":
                    SCENE[i]
            }
        )

pd.DataFrame(
    manifest
).to_csv(
    OUT /
    "patientwise_split_manifest.csv",
    index=False
)

# =============================================================================
# 14. 3D CNN FEATURE EXTRACTOR
# =============================================================================

print("\n" + "=" * 90)
print("STEP 6 — TRAINING SPECTRAL–SPATIAL 3D CNN")
print("=" * 90)


def build_cnn(
    input_shape
):

    inp = layers.Input(
        shape=input_shape
    )

    x = layers.Reshape(
        (
            input_shape[0],
            input_shape[1],
            input_shape[2],
            1
        )
    )(inp)

    # spectral-spatial
    x = layers.Conv3D(
        16,
        (3, 3, 7),
        padding="same",
        activation="relu"
    )(x)

    x = layers.BatchNormalization()(x)

    x = layers.MaxPool3D(
        (2, 2, 2)
    )(x)

    x = layers.Conv3D(
        32,
        (3, 3, 5),
        padding="same",
        activation="relu"
    )(x)

    x = layers.BatchNormalization()(x)

    x = layers.MaxPool3D(
        (2, 2, 2)
    )(x)

    x = layers.Conv3D(
        64,
        (3, 3, 3),
        padding="same",
        activation="relu"
    )(x)

    x = layers.BatchNormalization()(x)

    x = layers.GlobalAveragePooling3D()(x)

    # Feature representation
    features = layers.Dense(
        128,
        activation="relu",
        name="deep_features"
    )(x)

    x = layers.Dropout(
        0.40
    )(features)

    output = layers.Dense(
        3,
        activation="softmax",
        name="class_output"
    )(x)

    return models.Model(
        inp,
        output
    )


cnn = build_cnn(
    X.shape[1:]
)

cnn.compile(
    optimizer=tf.keras.optimizers.Adam(
        1e-3
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

cnn.summary()

callbacks = [

    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=4,
        restore_best_weights=True
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        patience=2,
        factor=0.5
    ),

    tf.keras.callbacks.ModelCheckpoint(
        OUT /
        "best_3DCNN.keras",
        monitor="val_loss",
        save_best_only=True
    )
]

history = cnn.fit(

    X[train_idx].astype(
        np.float32
    ),
    Y[train_idx],

    validation_data=(
        X[val_idx].astype(
            np.float32
        ),
        Y[val_idx]
    ),

    epochs=CNN_EPOCHS,

    batch_size=BATCH,

    callbacks=callbacks,

    verbose=1
)

# =============================================================================
# 15. CNN BASELINE
# =============================================================================

print("\n" + "=" * 90)
print("STEP 7 — CNN BASELINE")
print("=" * 90)

cnn_prob = cnn.predict(
    X[test_idx].astype(
        np.float32
    ),
    batch_size=BATCH,
    verbose=0
)

cnn_pred = np.argmax(
    cnn_prob,
    axis=1
)

cnn_acc = accuracy_score(
    Y[test_idx],
    cnn_pred
)

_, _, cnn_f1, _ = (
    precision_recall_fscore_support(
        Y[test_idx],
        cnn_pred,
        average="macro",
        zero_division=0
    )
)

print(
    "CNN Accuracy:",
    cnn_acc
)

print(
    "CNN Macro-F1:",
    cnn_f1
)

# =============================================================================
# 16. EXTRACT DEEP FEATURES
# =============================================================================

print("\n" + "=" * 90)
print("STEP 8 — EXTRACTING CNN FEATURES")
print("=" * 90)

feature_model = tf.keras.Model(
    inputs=cnn.input,
    outputs=cnn.get_layer(
        "deep_features"
    ).output
)

F_train = feature_model.predict(
    X[train_idx].astype(
        np.float32
    ),
    batch_size=BATCH,
    verbose=0
)

F_val = feature_model.predict(
    X[val_idx].astype(
        np.float32
    ),
    batch_size=BATCH,
    verbose=0
)

F_test = feature_model.predict(
    X[test_idx].astype(
        np.float32
    ),
    batch_size=BATCH,
    verbose=0
)

print(
    "CNN feature dimensions:",
    F_train.shape
)

# Free raw patches before quantum work
del X

gc.collect()

# =============================================================================
# 17. STANDARDIZATION + PCA FOR QUANTUM EMBEDDING
# =============================================================================

print("\n" + "=" * 90)
print("STEP 9 — PCA QUANTUM FEATURE REDUCTION")
print("=" * 90)

scaler = StandardScaler()

F_train_s = scaler.fit_transform(
    F_train
)

F_val_s = scaler.transform(
    F_val
)

F_test_s = scaler.transform(
    F_test
)

pca = PCA(
    n_components=PCA_COMPONENTS,
    random_state=SEED
)

Q_train = pca.fit_transform(
    F_train_s
)

Q_val = pca.transform(
    F_val_s
)

Q_test = pca.transform(
    F_test_s
)

print(
    "Quantum dimensions:",
    Q_train.shape[1]
)

print(
    "Explained variance:",
    pca.explained_variance_ratio_.sum()
)

# Scale PCA features to [-pi, pi]
q_scale = np.max(
    np.abs(Q_train),
    axis=0
)

q_scale[
    q_scale < 1e-8
] = 1.0


def angle_scale(z):

    z = z / q_scale

    z = np.clip(
        z,
        -1,
        1
    )

    return (
        z * np.pi
    ).astype(
        np.float64
    )


Q_train = angle_scale(
    Q_train
)

Q_val = angle_scale(
    Q_val
)

Q_test = angle_scale(
    Q_test
)

# =============================================================================
# 18. CLASS-BALANCED QUANTUM SUBSET
# =============================================================================

def balanced_subset(
    Xf,
    yf,
    maximum,
    seed
):

    rng = np.random.default_rng(
        seed
    )

    classes = np.unique(yf)

    per_class = max(
        1,
        maximum // len(classes)
    )

    chosen = []

    for c in classes:

        ids = np.where(
            yf == c
        )[0]

        n = min(
            len(ids),
            per_class
        )

        chosen.extend(
            rng.choice(
                ids,
                n,
                replace=False
            )
        )

    chosen = np.asarray(
        chosen
    )

    rng.shuffle(
        chosen
    )

    return (
        Xf[chosen],
        yf[chosen],
        chosen
    )


Y_train = Y[train_idx]
Y_val = Y[val_idx]
Y_test = Y[test_idx]

Q_train_sub, yq_train, _ = balanced_subset(
    Q_train,
    Y_train,
    MAX_Q_TRAIN,
    SEED
)

Q_val_sub, yq_val, _ = balanced_subset(
    Q_val,
    Y_val,
    MAX_Q_VAL,
    SEED + 1
)

Q_test_sub, yq_test, qtest_ids = balanced_subset(
    Q_test,
    Y_test,
    MAX_Q_TEST,
    SEED + 2
)

print(
    "\nQuantum train:",
    Q_train_sub.shape
)

print(
    "Quantum validation:",
    Q_val_sub.shape
)

print(
    "Quantum test:",
    Q_test_sub.shape
)

print(
    "Train classes:",
    Counter(yq_train)
)

# =============================================================================
# 19. CLASSICAL SVM BASELINE ON SAME FEATURES
# =============================================================================

print("\n" + "=" * 90)
print("STEP 10 — CLASSICAL RBF-SVM BASELINE")
print("=" * 90)

svm = SVC(
    kernel="rbf",
    C=10,
    gamma="scale",
    probability=True,
    random_state=SEED
)

svm.fit(
    Q_train_sub,
    yq_train
)

svm_pred = svm.predict(
    Q_test_sub
)

svm_acc = accuracy_score(
    yq_test,
    svm_pred
)

_, _, svm_f1, _ = (
    precision_recall_fscore_support(
        yq_test,
        svm_pred,
        average="macro",
        zero_division=0
    )
)

print(
    "CNN + PCA + SVM Accuracy:",
    svm_acc
)

print(
    "CNN + PCA + SVM Macro-F1:",
    svm_f1
)

# =============================================================================
# 20. QUANTUM CIRCUIT
# =============================================================================

print("\n" + "=" * 90)
print("STEP 11 — BUILDING VARIATIONAL QUANTUM CLASSIFIER")
print("=" * 90)

dev = qml.device(
    "default.qubit",
    wires=N_QUBITS
)


@qml.qnode(
    dev,
    interface="autograd"
)
def quantum_circuit(
    x,
    weights
):

    # Angle embedding
    qml.AngleEmbedding(
        x,
        wires=range(
            N_QUBITS
        ),
        rotation="Y"
    )

    # Variational layers
    qml.StronglyEntanglingLayers(
        weights,
        wires=range(
            N_QUBITS
        )
    )

    # Three outputs for L/N/P
    return [
        qml.expval(
            qml.PauliZ(i)
        )
        for i in range(3)
    ]


weight_shape = (
    VQC_LAYERS,
    N_QUBITS,
    3
)

from pennylane import numpy as pnp

weights = pnp.array(
    0.01 *
    np.random.randn(
        *weight_shape
    ),
    requires_grad=True
)

bias = pnp.array(
    np.zeros(3),
    requires_grad=True
)

print(
    "Qubits:",
    N_QUBITS
)

print(
    "Variational layers:",
    VQC_LAYERS
)

print(
    "Quantum parameters:",
    np.prod(weight_shape)
)

# =============================================================================
# 21. VQC FUNCTIONS
# =============================================================================

def logits_one(
    x,
    w,
    b
):

    qout = quantum_circuit(
        x,
        w
    )

    return pnp.stack(
        qout
    ) + b


def softmax(
    z
):

    z = z - pnp.max(z)

    e = pnp.exp(z)

    return e / pnp.sum(e)


def cross_entropy(
    logits,
    target
):

    prob = softmax(
        logits
    )

    return -pnp.log(
        prob[target]
        + 1e-10
    )


def batch_loss(
    w,
    b,
    Xb,
    yb
):

    losses = []

    for x, y in zip(
        Xb,
        yb
    ):

        logits = logits_one(
            x,
            w,
            b
        )

        losses.append(
            cross_entropy(
                logits,
                int(y)
            )
        )

    return pnp.mean(
        pnp.stack(losses)
    )


def vqc_predict(
    Xdata,
    w,
    b
):

    predictions = []

    probabilities = []

    for x in Xdata:

        logits = np.asarray(
            logits_one(
                x,
                w,
                b
            ),
            dtype=float
        )

        e = np.exp(
            logits
            - np.max(logits)
        )

        prob = e / e.sum()

        probabilities.append(
            prob
        )

        predictions.append(
            int(
                np.argmax(prob)
            )
        )

    return (
        np.asarray(predictions),
        np.asarray(probabilities)
    )

# =============================================================================
# 22. TRAIN VQC
# =============================================================================

print("\n" + "=" * 90)
print("STEP 12 — TRAINING VQC")
print("=" * 90)

optimizer = qml.AdamOptimizer(
    stepsize=LR_Q
)

history_q = []

best_val = -1

best_weights = None
best_bias = None

rng = np.random.default_rng(
    SEED
)

start = time.time()

for epoch in range(
    1,
    VQC_EPOCHS + 1
):

    order = rng.permutation(
        len(Q_train_sub)
    )

    losses = []

    for start_idx in range(
        0,
        len(order),
        VQC_BATCH
    ):

        ids = order[
            start_idx:
            start_idx + VQC_BATCH
        ]

        xb = Q_train_sub[
            ids
        ]

        yb = yq_train[
            ids
        ]

        weights, bias = optimizer.step(
            lambda w, b:
                batch_loss(
                    w,
                    b,
                    xb,
                    yb
                ),
            weights,
            bias
        )

        loss_value = float(
            batch_loss(
                weights,
                bias,
                xb,
                yb
            )
        )

        losses.append(
            loss_value
        )

    # Validation
    val_pred, _ = vqc_predict(
        Q_val_sub,
        weights,
        bias
    )

    val_acc = accuracy_score(
        yq_val,
        val_pred
    )

    _, _, val_f1, _ = (
        precision_recall_fscore_support(
            yq_val,
            val_pred,
            average="macro",
            zero_division=0
        )
    )

    epoch_loss = float(
        np.mean(losses)
    )

    history_q.append(
        {
            "epoch": epoch,
            "loss": epoch_loss,
            "val_accuracy":
                val_acc,
            "val_macro_f1":
                val_f1
        }
    )

    if val_f1 > best_val:

        best_val = val_f1

        best_weights = pnp.array(
            weights,
            requires_grad=False
        )

        best_bias = pnp.array(
            bias,
            requires_grad=False
        )

    print(
        f"Epoch {epoch:02d}/{VQC_EPOCHS} | "
        f"loss={epoch_loss:.4f} | "
        f"val_acc={val_acc:.4f} | "
        f"val_F1={val_f1:.4f}"
    )

quantum_time = (
    time.time()
    - start
)

print(
    "\nQuantum training time:",
    round(
        quantum_time / 60,
        2
    ),
    "minutes"
)

# =============================================================================
# 23. FINAL VQC TEST
# =============================================================================

print("\n" + "=" * 90)
print("STEP 13 — VQC TEST RESULTS")
print("=" * 90)

vqc_pred, vqc_prob = vqc_predict(
    Q_test_sub,
    best_weights,
    best_bias
)

vqc_acc = accuracy_score(
    yq_test,
    vqc_pred
)

vqc_bacc = balanced_accuracy_score(
    yq_test,
    vqc_pred
)

vqc_precision, vqc_recall, vqc_f1, _ = (
    precision_recall_fscore_support(
        yq_test,
        vqc_pred,
        average="macro",
        zero_division=0
    )
)

print(
    classification_report(
        yq_test,
        vqc_pred,
        target_names=CLASSES,
        zero_division=0
    )
)

print(
    "VQC Accuracy:",
    round(
        vqc_acc,
        4
    )
)

print(
    "VQC Balanced Accuracy:",
    round(
        vqc_bacc,
        4
    )
)

print(
    "VQC Macro-F1:",
    round(
        vqc_f1,
        4
    )
)

# =============================================================================
# 24. CONFUSION MATRIX
# =============================================================================

cm = confusion_matrix(
    yq_test,
    vqc_pred,
    labels=[
        0,
        1,
        2
    ]
)

plt.figure(
    figsize=(6, 5)
)

plt.imshow(cm)

plt.xticks(
    range(3),
    CLASSES
)

plt.yticks(
    range(3),
    CLASSES
)

plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "True"
)

plt.title(
    "CNN + VQC\n"
    "Patient-wise Test Set"
)

for i in range(3):

    for j in range(3):

        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )

plt.tight_layout()

plt.savefig(
    OUT /
    "VQC_confusion_matrix.png",
    dpi=300
)

plt.show()

# =============================================================================
# 25. QUANTUM LEARNING CURVE
# =============================================================================

qh = pd.DataFrame(
    history_q
)

qh.to_csv(
    OUT /
    "VQC_training_history.csv",
    index=False
)

plt.figure(
    figsize=(7, 5)
)

plt.plot(
    qh["epoch"],
    qh["val_accuracy"],
    label="Validation Accuracy"
)

plt.plot(
    qh["epoch"],
    qh["val_macro_f1"],
    label="Validation Macro-F1"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Score"
)

plt.ylim(
    0,
    1
)

plt.title(
    "Variational Quantum Classifier"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    OUT /
    "VQC_learning_curve.png",
    dpi=300
)

plt.show()

# =============================================================================
# 26. FINAL COMPARISON
# =============================================================================

print("\n" + "=" * 90)
print("STEP 14 — CLASSICAL vs QUANTUM COMPARISON")
print("=" * 90)

# CNN result is on complete patient-wise test set.
# SVM/VQC use same balanced quantum subset for fair direct comparison.

svm_bacc = balanced_accuracy_score(
    yq_test,
    svm_pred
)

svm_precision, svm_recall, svm_f1, _ = (
    precision_recall_fscore_support(
        yq_test,
        svm_pred,
        average="macro",
        zero_division=0
    )
)

results = pd.DataFrame(
    [
        {
            "Model":
                "3D-CNN",
            "Feature_Source":
                "Raw spectral-spatial patches",
            "Test_Set":
                "Full patient-wise test",
            "Accuracy":
                cnn_acc,
            "Macro_F1":
                cnn_f1
        },

        {
            "Model":
                "CNN + PCA + RBF-SVM",
            "Feature_Source":
                f"{N_QUBITS} CNN-PCA features",
            "Test_Set":
                "Balanced quantum test subset",
            "Accuracy":
                svm_acc,
            "Macro_F1":
                svm_f1
        },

        {
            "Model":
                "CNN + PCA + VQC",
            "Feature_Source":
                f"{N_QUBITS} qubits",
            "Test_Set":
                "Balanced quantum test subset",
            "Accuracy":
                vqc_acc,
            "Macro_F1":
                vqc_f1
        }
    ]
)

print(
    results.round(4).to_string(
        index=False
    )
)

results.to_csv(
    OUT /
    "R2_FINAL_RESULTS.csv",
    index=False
)

# =============================================================================
# 27. DIRECT SVM vs VQC FIGURE
# =============================================================================

direct = pd.DataFrame(
    {
        "Accuracy": [
            svm_acc,
            vqc_acc
        ],
        "Macro-F1": [
            svm_f1,
            vqc_f1
        ]
    },
    index=[
        "CNN+SVM",
        "CNN+VQC"
    ]
)

direct.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.ylim(
    0,
    1
)

plt.ylabel(
    "Score"
)

plt.title(
    "Classical vs Quantum Classifier\n"
    "Same Patient-wise Test Subset"
)

plt.xticks(
    rotation=0
)

plt.tight_layout()

plt.savefig(
    OUT /
    "Classical_vs_Quantum.png",
    dpi=300
)

plt.show()

# =============================================================================
# 28. SAVE VQC PREDICTIONS
# =============================================================================

test_original_indices = test_idx[
    qtest_ids
]

pred_df = pd.DataFrame(
    {
        "true_label": [
            CLASSES[i]
            for i in yq_test
        ],

        "predicted_label": [
            CLASSES[i]
            for i in vqc_pred
        ],

        "confidence":
            np.max(
                vqc_prob,
                axis=1
            ),

        "patient":
            GROUP[
                test_original_indices
            ],

        "scene":
            SCENE[
                test_original_indices
            ]
    }
)

for i, c in enumerate(
    CLASSES
):

    pred_df[
        f"prob_{c}"
    ] = vqc_prob[
        :, i
    ]

pred_df.to_csv(
    OUT /
    "VQC_test_predictions.csv",
    index=False
)

# =============================================================================
# 29. SAVE EXPERIMENT CONFIGURATION
# =============================================================================

config = {

    "research":
        "R2 Hybrid Quantum-Classical Spectral-Spatial Learning",

    "split":
        "Patient-wise",

    "patient_leakage":
        0,

    "patch_size":
        PATCH,

    "spectral_bands_used":
        N_BANDS,

    "cnn_feature_dimension":
        128,

    "pca_components":
        PCA_COMPONENTS,

    "qubits":
        N_QUBITS,

    "vqc_layers":
        VQC_LAYERS,

    "quantum_embedding":
        "AngleEmbedding-Y",

    "variational_circuit":
        "StronglyEntanglingLayers",

    "quantum_train_samples":
        int(
            len(Q_train_sub)
        ),

    "quantum_test_samples":
        int(
            len(Q_test_sub)
        ),

    "quantum_training_minutes":
        quantum_time / 60,

    "pca_explained_variance":
        float(
            pca.explained_variance_ratio_.sum()
        ),

    "CNN_accuracy":
        float(cnn_acc),

    "SVM_accuracy":
        float(svm_acc),

    "VQC_accuracy":
        float(vqc_acc),

    "VQC_macro_f1":
        float(vqc_f1)
}

with open(
    OUT /
    "experiment_configuration.json",
    "w"
) as f:

    json.dump(
        config,
        f,
        indent=4
    )

# =============================================================================
# 30. SAVE MODEL PARAMETERS
# =============================================================================

np.save(
    OUT /
    "VQC_weights.npy",
    np.asarray(
        best_weights
    )
)

np.save(
    OUT /
    "VQC_bias.npy",
    np.asarray(
        best_bias
    )
)

np.save(
    OUT /
    "PCA_components.npy",
    pca.components_
)

# =============================================================================
# 31. ZIP EVERYTHING
# =============================================================================

zip_base = (
    "/kaggle/working/"
    "R2_Hybrid_Quantum_HSI_Results"
)

shutil.make_archive(
    zip_base,
    "zip",
    OUT
)

# =============================================================================
# 32. FINAL REPORT
# =============================================================================

print("\n")
print("=" * 90)
print("R2 EXPERIMENT COMPLETE")
print("=" * 90)

print("\nArchitecture:")
print(
    "HSI -> 32 spectral bands -> "
    "25x25 spectral-spatial patches -> "
    "3D-CNN -> 128 features -> "
    f"PCA {N_QUBITS}D -> "
    f"{N_QUBITS}-qubit VQC -> L/N/P"
)

print("\nLeakage control:")
print(
    "PATIENT-WISE SPLIT — ZERO PATIENT OVERLAP"
)

print("\nFinal results:")
print(
    results.round(4).to_string(
        index=False
    )
)

print("\nFiles saved in:")
print(OUT)

print("\nZIP file:")
print(
    "/kaggle/working/"
    "R2_Hybrid_Quantum_HSI_Results.zip"
)

print("\nImportant outputs:")
print(
    "  R2_FINAL_RESULTS.csv"
)
print(
    "  patientwise_split_manifest.csv"
)
print(
    "  VQC_test_predictions.csv"
)
print(
    "  VQC_confusion_matrix.png"
)
print(
    "  VQC_learning_curve.png"
)
print(
    "  Classical_vs_Quantum.png"
)
print(
    "  VQC_weights.npy"
)
print(
    "  experiment_configuration.json"
)

print("=" * 90)
```
